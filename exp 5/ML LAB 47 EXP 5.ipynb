# Disease Prediction using Naive Bayes and Neural Network with Comparison of Classifiers

```python


from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split

from sklearn.linear_model import LogisticRegression 
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier

from sklearn.naive_bayes import GaussianNB
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score,roc_auc_score,confusion_matrix,classification_report

```

```python
data = load_breast_cancer()
x=data.data
y=data.target

print("classes:",data.target_names)

x_train, x_test, y_train, y_test = train_test_split(
    x,y,test_size=0.3)
```

```python
lr=LogisticRegression(max_iter=5000)
lr.fit(x_train,y_train)

pred_lr = lr.predict(x_test)

print("Accuracy_test:",accuracy_score (y_test,pred_lr))
pred_lr_train=lr.predict(x_train)
print("Accuracy_train:",accuracy_score (y_train,pred_lr_train))
```

```python
dt= DecisionTreeClassifier()
dt.fit(x_train,y_train)

pred_dt = dt.predict(x_test)

print("Accuracy_test:",accuracy_score (y_test,pred_dt))
pred_dt_train=dt.predict(x_train)
print("Accuracy_train:",accuracy_score (y_train,pred_dt_train))
```

```python
kn= KNeighborsClassifier()
kn.fit(x_train,y_train)

pred_kn = kn.predict(x_test)

print("Accuracy_test:",accuracy_score (y_test,pred_kn))
pred_kn_train=kn.predict(x_train)
print("Accuracy_train:",accuracy_score (y_train,pred_kn_train))
```

```python
gnb=GaussianNB()
gnb.fit(x_train,y_train)

pred_gnb=gnb.predict(x_test)
print("accuracy_test:",accuracy_score(y_test,pred_gnb))
pred_gnb_train=gnb.predict(x_train)
print("Accuracy_train:",accuracy_score (y_train,pred_gnb_train))
```

```python
mlp= MLPClassifier()
mlp.fit(x_train,y_train)

pred_mlp=mlp.predict(x_test)
print("accuracy_test:",accuracy_score(y_test,pred_mlp))
pred_mlp_train=mlp.predict(x_train)
print("Accuracy_train:",accuracy_score (y_train,pred_mlp_train))
```
